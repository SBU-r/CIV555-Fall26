## Operations

Review the most commonly used matrix operations.

### Transpose

```python
A = np.arange(1.,7.).reshape(2,3)
print(f'A=\n{A}\n')

A_T = np.transpose(A) 
print(f'The transpose of A, A^T=\n {A_T} \n')
```

### Addition & Subtraction

```python
A = np.arange(1.,7.).reshape(3,2)
print(f'A=\n {A} \n')

B = np.arange(7.,13.).reshape(3,2)
print(f'B=\n {B} \n')

d= np.arange(10.,13.).reshape(3,1)
print(f'd=\n {d} \n')

print(f'A+B=\n {A+B} \n')

print(f'2*A+3=\n {2*A+3} \n')

print(f'A+d=\n {A+d} \n')

print(f'A-B=\n {A-B} \n')
```

### Multiplication

```python
A = np.arange(1.,7.).reshape(3,2)
print(f'A=\n {A,}, \n')

B = np.arange(7.,13.).reshape(3,2)
print(f'B=\n {B} \n')

c = np.arange(4.,6.)
print(f'c=\n {c} \n')

d = np.arange(1.,4.)
print(f'd=\n {d} \n')

E = np.arange(1.,7.).reshape(2,3)
print(f'E=\n {E} \n')

# Hadamard product (element-wise product *)
print(f'A*B=\n {A*B} \n')

# matrix multiplication (using np.matmul)
print(f'dA=\n {np.matmul(d,A)} \n')

print(f'EA=\n {np.matmul(E,A)} \n')

print(f'Ac^T= {np.matmul(A,np.transpose(c))} \n')
```

### Matrix Diagonal

```python
A = np.arange(1.,17.).reshape(4,4)
print(f'A=\n {A} \n')

A_diag=np.diag(A) # find the diagonal of the matrix using np.diag()
print(f'Diagonal values of A:\n {A_diag}')
```

### Identity Matrix and Inverse Operator

```python
a = np.arange(2.,5.)
print(f'a=\n {a} \n')

I = np.eye(3)
print(f'I=\n {I} \n')

print(f'aI=\n {np.matmul(a,I)} \n') # check if aI does not change the vector a

print(f'Ia=\n {np.matmul(I,np.transpose(a))}') # check if Ia does not change a
```

```python
# Inverse of a matrix
A = np.arange(1.,5.).reshape(2,2)
print(f'A=\n {A} \n')

A_inv=np.linalg.inv(A) # find the inverse of a matrix using np.linalg.inv()
print(f'The inverse of A is, \n {A_inv} \n')

print(f'A multiplied by its inverse matrix, AA_inv, is the identity matrix:\n {np.matmul(A,A_inv)} \n') # the product of A and its inverse is the identity matrix
```

### Trace

```python
A = np.arange(1.,10.).reshape(3,3)
print(f'A=\n {A} \n')

A_trace = np.trace(A) # np.trace() is the function for finding the trace of a matrix
print(f'The trace of A is: \n {A_trace}') # check whether A_trace is the sum of the diagonal elements
```

### Determinant

```python
A=np.arange(1.,5.).reshape(2,2)
print(f'A:\n {A} \n')

A_det=np.linalg.det(A) # find the determinant of a matrix using np.linalg.det()
print(f'determinant of A:\n {A_det:.2f}')
```

## Norms

`np.linalg.norm()` computes different norms for vectors and matrices.

### $l_1$-Norm

```python
A=np.arange(-2.,2.)
print(f'A:\n {A} \n')

A_norm1=np.linalg.norm(A,ord=1) #np.linalg.norm(tensor, ord=) computes the norm of a tensor
print(f'L_1 norm of A:\n {A_norm1:.2f}')
```

### $l_2$-Norm

```python
A=np.arange(-2.,2.)
print(f'A:\n {A} \n')

A_norm2=np.linalg.norm(A,ord=2)
print(f'L_2 norm of A:\n {A_norm2:.2f}')
```

### $l_\infty$-Norm

```python
A=np.arange(-2.,2.)
print(f'A:\n {A} \n')

A_norminf=np.linalg.norm(A,ord=np.inf)
print(f'L_inf norm of A:\n {A_norminf:.2f}')
```

### Frobenius Norm

```python
A=np.arange(-2.,2.).reshape(2,2)
print(f'A:\n {A} \n')

A_Fro=np.linalg.norm(A,ord='fro')
print(f'Frobenius norm of A:\n {A_Fro:.2f}')
```

## Decompositions

### Eigen Decomposition

```python
A = np.array([5.,1.,3.,3.]).reshape(2,2)
print(f'A:\n {A} \n')

e,V=np.linalg.eig(A)# find eigenvalues and eigenvectors of a matrix using np.linalg.eig()
print(f'eigenvalues of A:\n e = {e} \n')

print(f'eigenvectors of A: \n V = {V} \n')

L=(np.diag(e)) # construct the eigenvalue matrix using np.diag()
print(f'Eigenvalue matrix:\n {L} \n')

print(f'EigenDecomposition of A:\n {abs(np.matmul(np.matmul(V,L),np.linalg.inv(V)))-A} \n') # check whether A equals its eigendecomposition VLV^(-1)


print(f'det(A)-the product of eigenvalues:\n {np.linalg.det(A)-np.prod(e):.3f} \n') # check whether the determinant of A equals the product of the eigenvalues
```

### Singular Value Decomposition

```python
# define a matrix A 
A = np.random.randn(2,3)
print(f'A:\n {A} \n')
```

```python
# singular value decomposition using np.linalg.svd()
U,D,V = np.linalg.svd(A, full_matrices=False) 
print(f'left singular vectors, U:\n {U} \n')
print(f'singular values, D:\n {D} \n')
print(f'transpose of right singular vectors, V:\n {V} \n')

print(f'right singular vectors, V:\n {np.transpose(V)} \n')

# let's check whether UDV is approximately equal to A
print(f'UDV-A:\n {np.matmul(np.matmul(U,np.diag(D)), V)-A}') 

```

```python
# Let's check whether the left-singular vectors of A are the eigenvectors of AA^T 
L,V=np.linalg.eig(np.matmul(A, np.transpose(A)))
print(f'eigenvectors of AA^T:\n {V} \n')

# and whether the square root of the non-zero eigenvalues of AA^T are the singular values of A
print(f'square root of nonzero eigenvalues of AA^T:\n {np.sqrt(L)} \n')
```

```python
# Let's check whether the right-singular vectors of A are the eigenvectors of A^TA and whether the square root of the nonzero singular values of A are the eigenvalues of A^TA
L,V=np.linalg.eig(np.matmul(np.transpose(A),A))

print(f'eigenvectors of A^TA:\n {V} \n')

print(f'square root of nonzero eigenvalues of A^TT:\n {np.sqrt(L)} \n')
```
